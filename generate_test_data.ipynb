# Generate `test_data.csv`
Creates the 20% stratified hold-out set from the UCI Adult dataset (same cleaning and seed 42 as the main assignment notebook). Works in Google Colab or any Jupyter environment – no local Python needed.

Run all cells, then download `test_data.csv` with the last cell.

```python
import os
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

SEED = 42
COLUMNS = ["age", "workclass", "fnlwgt", "education", "education_num", "marital_status",
           "occupation", "relationship", "race", "sex", "capital_gain", "capital_loss",
           "hours_per_week", "native_country", "income"]
URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/adult/adult.data"
LOCAL = "data/adult.csv"
os.makedirs("data", exist_ok=True)

if os.path.exists(LOCAL):
    df = pd.read_csv(LOCAL)
else:
    df = pd.read_csv(URL, header=None, names=COLUMNS, skipinitialspace=True, na_values="?")
    df.to_csv(LOCAL, index=False)
df.columns = COLUMNS
df = df.replace("?", np.nan)
df["income"] = (df["income"].astype(str).str.strip().str.rstrip(".") == ">50K").astype(int)
df = df.drop_duplicates().reset_index(drop=True)
print("Cleaned dataset:", df.shape)
```

```python
X, y = df.drop(columns="income"), df["income"]
_, X_test, _, y_test = train_test_split(X, y, test_size=0.20, stratify=y, random_state=SEED)
test_df = pd.concat([X_test, y_test], axis=1)
test_df.to_csv("test_data.csv", index=False)
print("Saved test_data.csv with", len(test_df), "rows")
test_df.head()
```
